# While Loops & De Math Library in Python
In deze les gaan we leren hoe we acties kunnen herhalen zolang een voorwaarde **waar (`True`)** is met behulp van de `while`-loop. We combineren dit met de ingebouwde **`math`-bibliotheek** voor wiskundige berekeningen.

## Deel 1: De Basis van de `while`-loop
Een `for`-loop gebruik je als je van tevoren weet *hoe vaak* je iets wilt herhalen. Een `while`-loop gebruik je als je wilt herhalen *zolang een voorwaarde geldt*.

```python
# Eenvoudige teller met een while-loop
teller = 1

while teller <= 5:
    print(f"Teller staat op: {teller}")
    teller += 1  # BELANGRIJK: vergeet de teller niet op te hogen!

print("Loop is afgelopen!")
```

## Deel 2: De `math` Library
Python heeft een ingebouwde bibliotheek voor wiskundige functies. Je moet deze eerst importeren met `import math`.

```python
import math

# Voorbeelden van math functies
print("Wortel van 25:", math.sqrt(25))
print("Het getal pi:", math.pi)
print("Afronden naar boven (3.2):", math.ceil(3.2))
print("Afronden naar beneden (3.8):", math.floor(3.8))
print("5 faculteit (5!):", math.factorial(5))
```

## Deel 3: `while`-loop combineren met `math`
Stel dat we willen weten bij welk getal $x$ de wortel $\sqrt{x}$ groter wordt dan $10$.

```python
import math

getal = 1
while math.sqrt(getal) <= 10:
    getal += 1

print(f"Het eerste getal waarvan de wortel groter is dan 10 is: {getal}")
print(f"Wortel van {getal} = {math.sqrt(getal)}")
```
