# Week 2 Huiswerk: De Drone-vluchtanalyzer
In deze opdracht ga je een vliegpad analyseren dat bestaat uit meerdere GPS-waypoint-coördinaten. Elk waypoint is een lijst met `[Breedtegraad, Lengtegraad, Hoogte]`.

### De Data
Kopieer deze lijst naar je codecel:
```python
flight_path = [
    [52.3702, 4.8952, 10],  # Takeoff
    [52.3710, 4.8960, 45],  
    [52.3725, 4.8975, 125], # High altitude
    [52.3740, 4.8990, 135], # High altitude
    [52.3755, 4.9005, 30]   # Landing approach
]
```

--- 
### Je Takenteller:
1. **Doorloop** de `flight_path` met behulp van een `for` loop.
2. **Extraheer** de hoogte (de 3e waarde) voor elk waypoint.
3. **Controleer** of de hoogte boven de **120 meter** ligt (de wettelijke limiet voor veel drones).
4. **Tel** hoeveel waypoints illegaal waren.
5. **Print** een samenvatting aan het einde: `"Total illegal waypoints found: [count]"`.

````python
flight_path = [
    [52.3702, 4.8952, 10], 
    [52.3710, 4.8960, 45],  
    [52.3725, 4.8975, 125], 
    [52.3740, 4.8990, 135], 
    [52.3755, 4.9005, 30]   
]

illegal_count = 0

# Start your loop here...
````

  Gebruik `waypoint[2]` om de hoogte van een waypoint binnen de loop op te halen. 
  Gebruik in de loop een `if` statement om te controleren of de hoogte > 120 is, en tel in dat geval 1 op bij `illegal_count`.

  ```python
  for pt in flight_path:
      altitude = pt[2]
      if altitude > 120:
          print(f"Violation detected at {altitude}m!")
          illegal_count += 1
  
  print(f"Total illegal waypoints found: {illegal_count}")
  ```
