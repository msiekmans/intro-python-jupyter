## 1. Lijsten (De Datacontainers)
Met een **Lijst** (List) kun je meerdere items in één enkele variabele opslaan met behulp van rechte haakjes `[]`. In Geomaps gebruiken we deze om coördinaten, stadsnamen of hoogtedata op te slaan.

### Kernbegrippen:
* **Aanmaken:** `my_list = ["Item1", "Item2", "Item3"]`
* **Indexering:** Python begint te tellen bij **0**.
    * `coords = [52.37, 4.89]` 
    * `coords[0]` is de Breedtegraad (52.37)
    * `coords[1]` is de Lengtegraad (4.89)
* **Data toevoegen:** Gebruik `.append()` om een nieuw item aan het einde van een lijst toe te voegen.
    * `cities.append("Groningen")`
* **Tellen:** Gebruik `len(list_name)` om te zien hoeveel items er in je dataset zitten.

---

## 2. For Loops (De Automatiseringsmotor)
Een **For Loop** herhaalt een actie voor elk item in een lijst. Zie het als een **lopende band**: hij pakt één item op, voert een taak uit en gaat automatisch door naar het volgende item totdat de lijst leeg is.

### De Basissyntaxis:
```python
cities = ["Amsterdam", "Rotterdam", "Utrecht"]

for city in cities:
    # Alles wat hier inspringt wordt uitgevoerd voor elke stad
    print(f"Mapping the city of: {city}")
```

---

## 3. De range() Functie

De **range()** functie genereert een reeks getallen zonder dat je vooraf een lijst hoeft te maken.  
```python
range(1, 6) 
```
Dit genereert: 1, 2, 3, 4, 5 (het stopt vóór het laatste getal).  
**Toepassing:** Ideaal voor het maken van unieke ID-nummers voor kaartmarkers of het een vast aantal keer herhalen van een berekening.

## 4. Lijst van Lijsten
We kunnen een "Lijst van Lijsten" (een route) analyseren:

### De Syntaxis:
```python
# Een lijst van [Lat, Lon, Alt]
route = [
    [52.1, 4.5, 10], 
    [52.2, 4.6, 130], 
    [52.3, 4.7, 20]
]

for point in route:
    altitude = point[2]  # Toegang tot het 3e item (index 2)
    if altitude > 120:
        print(f"Alert: Altitude {altitude}m exceeds legal limit!")

```
