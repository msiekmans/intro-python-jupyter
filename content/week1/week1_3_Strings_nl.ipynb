# Week 1, Deel 3: Werken met tekst (Strings)
In GIS en Data Science werken we veel met namen, categorieën en labels. In Python noemen we dit **Strings**.

### Opdracht 1: De begroeting
Maak de variabelen `first_name` en `last_name` aan. Gebruik een **f-string** om het volgende te printen: `"Welcome, [First Name] [Last Name]!"`

```python
first_name = "John"
last_name = "Doe"
print(f"Welcome, {first_name} {last_name}!")
```

### Opdracht 2: Data opschonen (Hoofdletters)
Maak een variabele `city = "utrecht"`. Gebruik de `.upper()` methode om dit af te drukken als `UTRECHT`.

```python
city = "utrecht"
print(city.upper())
```

### Opdracht 3: De adresbouwer
Maak `house_number = 10` (geheel getal / integer) en `street = "Dam"` (string). Gebruik een f-string om ze te combineren in één enkele variabele genaamd `full_address`.

```python
house_number = 10
street = "Dam"
full_address = f"{street} {house_number}"
print(full_address)
```

### Opdracht 4: Stringlengte
Gebruik de `len()` functie om het aantal tekens in de volgende string te tellen: `"Geographic Information Systems"`.

```python
text = "Geographic Information Systems"
length = len(text)
print(f"The text is {length} characters long.")
```

### Opdracht 5: Zoeken in tekst
Controleer of het woord `"river"` voorkomt in de zin: `"The Rhine is a major river in Europe"`. (Het resultaat moet True zijn).

```python
sentence = "The Rhine is a major river in Europe"
check1 = "river" in sentence
check2 = "Netherlands" in sentence
print(check1)
print(check2)
```

## Interactieve programma's
We gebruiken de `input()` functie om de gebruiker informatie aan ons programma te laten geven.

**Belangrijk:** `input()` slaat gegevens altijd op als **Tekst (String)**. Als je er berekeningen mee wilt uitvoeren, moet je deze omzetten met `int()` of `float()`.

### Opdracht 6: De Welkomstbot
Vraag de gebruiker om hun naam en print: `"Hello [naam], welcome to the GIS lab!"`

```python
user_name = input("Please enter your name: ")
print(f"Hello {user_name}, welcome to the GIS lab!")
```

### Opdracht 7: De Wiskundige Omrekenaar
Vraag de gebruiker om een afstand in **kilometer**. Zet dit om naar een **float**, vermenigvuldig het met 1000 en print de afstand in **meter**.

```python
km_text = input("How many kilometers did you walk? ")
km_value = float(km_text)
meters = km_value * 1000
print(f"That is exactly {meters} meters.")
```

### Opdracht 8: Professionele Labels
Vraag de gebruiker om een `city` en een `country`. Gebruik een f-string om ze af te drukken als een officiële kaarTtitel: `"CITY, COUNTRY"` (bijv. "AMSTERDAM, NETHERLANDS").

*Tip: Gebruik de .upper() methode!*

```python
city = input("In which city are you now? ")
country = input("In what country is that? ")
title = f"{city.upper()}, {country.upper()}"
print(title)
```
