## Opdracht 10: Formule van Heron
**Doel:** Alles combineren (Machten, Haakjes en Meerdere Bewerkingen).

**Vraag:** Vind de oppervlakte van een driehoek met zijden **a=50**, **b=120**, en **c=130**.

**Stap 1:** Bereken de halve omtrek $s$:
$$s = (a + b + c) / 2$$

**Stap 2:** Bereken de oppervlakte met behulp van de formule van Heron:
$$\text{Oppervlakte} = \sqrt{s(s-a)(s-b)(s-c)}$$

*Tip: Om een vierkantswortel te berekenen in Python, verhef je de gehele expressie tot de macht 0.5 met `** 0.5`.*

<details>
    <summary>Oplossing</summary>
    De berekening moet uitkomen op: 3000 
</details>
