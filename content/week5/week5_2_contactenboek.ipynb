## Deel 1: De basis van een Dictionary
Een **dictionary** slaat gegevens op in paren van een **key** (sleutel) en een **value** (waarde). Dit is handiger dan een `list`, omdat je niet op indexnummers hoeft te zoeken, maar op duidelijke namen.

```python
# Voorbeeld van een enkele dictionary
contact = {
    "naam": "Jan Jansen",
    "telefoon": "06-12345678",
    "email": "jan@example.com"
}

# Gegevens ophalen via de key
print("Naam:", contact["naam"])
print("Telefoon:", contact["telefoon"])

# Waarde aanpassen
contact["telefoon"] = "06-99999999"

# Nieuwe key-value toevoegen
contact["stad"] = "Amsterdam"

print("\nBijgewerkte dictionary:", contact)
```

## Deel 2: Geneste Dictionaries (Nested Dictionaries)

Wat als we **meerdere contacten** willen opslaan? Dan kunnen we een dictionary maken waarin de waarde *zelf ook weer een dictionary is*!

```python
# Geneste dictionary
contacten = {
    "Jan Jansen": {
        "telefoon": "06-12345678",
        "email": "jan@example.com",
        "stad": "Amsterdam"
    },
    "Sanne de Vries": {
        "telefoon": "06-87654321",
        "email": "sanne@example.com",
        "stad": "Utrecht"
    }
}

# Specifieke waarde ophalen met dubbele brackets [outer_key][inner_key]
print("Stad van Sanne:", contacten["Sanne de Vries"]["stad"])

# Nieuw contact toevoegen
contacten["Lotte Bakker"] = {
    "telefoon": "06-55512345",
    "email": "lotte@example.com",
    "stad": "Rotterdam"
}

# Door geneste dictionary loopen
print("\n--- Alle Contacten ---")
for naam, details in contacten.items():
    print(f"{naam} woont in {details['stad']} (Tel: {details['telefoon']})")
```
