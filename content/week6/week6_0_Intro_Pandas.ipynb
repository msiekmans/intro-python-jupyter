# Introductie Pandas - DataFrames, Datatypen & CSV
In deze les maken we kennis met **Pandas**, de belangrijkste Python-bibliotheek voor databewerking en -analyse. We leren hoe we zelf handmatig DataFrames maken, datatypen beheren en hoe we gegevens exporteren naar en inlezen uit **CSV-bestanden**.

## Deel 1: Handmatig een DataFrame maken
Een **DataFrame** in Pandas is een 2D-tabel met rijen en kolommen (vergelijkbaar met een Excel-werkblad of SQL-tabel). De meest gebruikelijke manier om handmatig een DataFrame te maken is met een Python **dictionary**.

```python
import pandas as pd

# Handmatig data aanmaken met een dictionary
data = {
    "Werknemer": ["Jan", "Sanne", "Lotte", "Peter"],
    "Afdeling": ["IT", "HR", "Finance", "IT"],
    "Salaris": [4200, 3800, 4500, 3900],
    "In_Dienst": [True, True, False, True]
}

# Omzetten naar een Pandas DataFrame
df = pd.DataFrame(data)

# Bekijk de eerste rijen
df
```

## Deel 2: Datatypen controleren en aanpassen
Pandas kent automatisch datatypen toe aan kolommen (zoals `int64`, `float64`, `bool`, of `object` voor tekst). Met `.dtypes` of `.info()` kun je dit controleren en met `.astype()` kun je dit aanpassen.

```python
# Controleren van datatypen
print("--- Datatypen van de kolommen ---")
print(df.dtypes)

# Datatype van 'Salaris' aanpassen van integer naar float
df["Salaris"] = df["Salaris"].astype(float)

# Datum toevoegen en omzetten naar het juiste datetime datatype
df["Startdatum"] = ["2020-01-15", "2019-05-01", "2021-11-01", "2022-03-10"]
df["Startdatum"] = pd.to_datetime(df["Startdatum"])

print("
--- Bijgewerkte Datatypen ---")
print(df.dtypes)
df
```

## Deel 3: Exporteren naar CSV (`to_csv`)
Als je DataFrame klaar is, kun je deze opslaan als CSV-bestand. Gebruik altijd `index=False` om te voorkomen dat Pandas de automatische rijnummers als een losse kolom opslaat.

```python
# Exporteren naar een CSV-bestand
df.to_csv("werknemers.csv", index=False, sep=",")
print("Bestand 'werknemers.csv' is succesvol opgeslagen!")
```

## Deel 4: CSV inlezen (`read_csv`)
Met `pd.read_csv()` kun je een opgeslagen CSV-bestand weer inlezen als DataFrame.

```python
# Inlezen vanuit CSV
nieuwe_df = pd.read_csv("werknemers.csv")

# Bekijk de ingelezen data
print("Ingelezen DataFrame:")
nieuwe_df.head()
```
