# Week 3: Vergelijkingsoperators & Besturingsstructuren

In dit practicum leren we hoe we Python kunnen laten "nadenken". We gebruiken **Vergelijkingsoperators** om data te evalueren en **If/Else-statements** om verschillende acties uit te voeren op basis van die resultaten.

--- 

## 1. Logica & Vergelijkingsoperators
Vergelijkingsoperators vergelijken twee waarden en geven een **Boolean** (`True` of `False`) terug. 

| Operator | Betekenis | Voorbeeld |
| :--- | :--- | :--- |
| `==` | Gelijk aan | `city == "Amsterdam"` |
| `!=` | Niet gelijk aan | `status != "Error"` |
| `>` | Groter dan | `elevation > 100` |
| `<` | Kleiner dan | `temp < 0` |
| `>=` | Groter of gelijk aan | `battery >= 20` |
| `<=` | Kleiner of gelijk aan | `dist <= 500` |

--- 

## 2. If / Elif / Else (Besturingsstructuren)
Deze structuur werkt als een beslissingsboom. 

**Cruciale Regel:** Python gebruikt **Inspringing (4 spaties)** om te weten welke code bij de beslissing hoort.

```python
dist_to_river = 450

if dist_to_river < 500:
    print("High Risk Zone")
elif dist_to_river <= 1000:
    print("Medium Risk Zone")
else:
    print("Safe Zone")
```
