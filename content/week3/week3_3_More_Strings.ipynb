## Deel 1: Strings Vergelijken (`==` vs Java `.equals()`)
In Python vergelijk je de waarde van twee strings met de operator **`==`**. Anders dan in Java heb je hier geen `.equals()` voor nodig.

```python
woord1 = "python"
woord2 = "python"
woord3 = "Python"

print("woord1 == woord2:", woord1 == woord2)  # True
print("woord1 == woord3:", woord1 == woord3)  # False
print("Case-insensitive:", woord1.lower() == woord3.lower())  # True
```

## Deel 2: String Slicing (Deelsnijden)
Met slicing kun je specifieke karakters of gedeeltes uit een string halen met de syntaxis `string[start:stop:stap]`.

```python
tekst = "JupyterNotebook"
print("Eerste 7 tekens:", tekst[:7])        # 'Jupyter'
print("Vanaf index 7:", tekst[7:])          # 'Notebook'
print("Omgedraaide tekst:", tekst[::-1])     # 'koobetoNretypuJ'
```

## Deel 3: Splitting & Joining (Explode & Implode)
Met `.split()` breek je een tekst op in een lijst (in Java/PHP vaak *explode* genoemd). Met `.join()` voeg je een lijst weer samen.

```python
zin = "Jan,Piet,Klaas,Sanne"
namen = zin.split(",")
print("Split:", namen)
print("Join:", " - ".join(namen))
```
